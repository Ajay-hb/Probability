```python
import math
def Gaussian_cdf(data):
    sorted_data=np.sort(data)
    CDF_values=[]
    mean=np.mean(sorted_data)
    sd=np.std(sorted_data)
    for i in sorted_data:
        k=(i-mean)/sd*np.sqrt(2)
        erf_value=math.erf(k)
        final_result=1/2*(1+erf_value)
        CDF_values.append(final_result)
    return CDF_values
```

```python
import numpy as np
def permutation_test(sample1,sample2,no_of_resamples):
    sample1_Array=np.array(sample1)
    sample2_Array=np.array(sample2)
    D=abs(np.mean(sample1_Array)-np.mean(sample2_Array))
    combined_sample=np.concatenate((sample1_Array,sample2_Array))
    n1=len(sample1_Array)
    n2=len(combined_sample)
    diff_lst=[]
    for i in range(no_of_resamples):
        jumbled_sample=np.random.permutation(combined_sample)
        new_sample1=jumbled_sample[:n1]
        new_sample2=jumbled_sample[n1:n2]
        diff=abs(np.mean(new_sample1)-np.mean(new_sample2))
        diff_lst.append(diff)
    diff_array=np.array(diff_lst)
    diff_array_sorted=np.sort(diff_array)
    count=0
    for i in diff_array_sorted:
        if i<D:
            count+=1
    percentage=(count/len(diff_array_sorted))*100
    return percentage
```

```python
sample1=np.random.randint(140,192,50)
sample2=np.random.randint(140,192,50)
no_of_resamples=1000
p_value=permutation_test(sample1,sample2,no_of_resamples)
```

```python
import numpy as np
import math
def GCDF(data):
    sorted_data=np.sort(data)# sort the data
    mean=np.mean(data)#mean
    std=np.std(data)#standard deviation
    GCDF_values=[]
    for i in sorted_data:
        k=(i-mean)/(np.sqrt(2)*std)
        erf=math.erf(k)
        final=1/2*(1+erf)
        GCDF_values.append(final)
    GCDF_array=np.array(GCDF_values)
    return GCDF_array
def ECDF(data):
    sorted_data=sorted(data)
    ECDF_values=[]
    n=len(sorted_data)
    i=0
    while i<n:
        var=(i+1)/n
        ECDF_values.append(var)
        i+=1 # Fixed: Increment i here
    ECDF_array=np.array(ECDF_values)
    return ECDF_array
def kstest(data):
    gcdf_values=GCDF(data)
    ecdf_values=ECDF(data)
    n=len(data)
    i=0
    diff=0
    while i<n:
        value1=gcdf_values[i]
        value2=ecdf_values[i]
        D=abs(value1-value2)
        if D>diff:
            diff=D
        i+=1 # Fixed: Increment i here
    critical_value=1.385*np.sqrt((n+n)/(n*n))
    return diff,critical_value
```

```python
import numpy as np
data=np.random.normal(0,1,100)
test_stats,critical=kstest(data) # Commented out to prevent infinite loop
```

```python
import math
def Gaussian_cdf_height(data):
  sorted_data=np.sort(data) # Sort the input data in ascending order
  GCDF_values=[] # Initialize an empty list to store the calculated CDF values
  mean=np.mean(sorted_data) # Calculate the mean of the sorted data
  sd=np.std(sorted_data) # Calculate the standard deviation of the sorted data
  for i in sorted_data:
    # Standardize and scale the data point using the formula for the normal CDF related to the error function
    k=(i-mean)/(np.sqrt(2)*sd)
    erf=math.erf(k) # Calculate the error function value for 'k'
    final=1/2*(1+erf) # Apply the formula to get the Gaussian CDF value
    GCDF_values.append(final) # Add the calculated CDF value to the list
  GCDF_Array=np.array(GCDF_values) # Convert the list of CDF values to a NumPy array
  return GCDF_Array # Return the array of Gaussian CDF values
```

```python
def empirical_bootstrapping_height(data,no_of_samples,sample_size,CI):
  median_list=[] # Initialize an empty list to store medians of bootstrap samples
  for i in range(no_of_samples):
    # Generate a bootstrap sample by randomly choosing from the original data with replacement
    sample=np.random.choice(data,size=sample_size,replace=True)
    median_list.append(np.median(sample)) # Calculate the median of the current bootstrap sample and add it to the list
  median_array=np.array(median_list) # Convert the list of medians to a NumPy array
  sorted_median_Array=np.sort(median_array) # Sort the array of medians

  # Calculate the lower and upper percentile ranks for the confidence interval
  LL_per=(100-CI)/2 # Lower percentile (e.g., for 95% CI, this is 2.5%)
  UL_per=(100-LL_per) # Upper percentile (e.g., for 95% CI, this is 97.5%)

  # Find the values at the calculated percentiles to determine the confidence interval bounds
  LL=np.percentile(sorted_median_Array,LL_per) # Lower Limit of the confidence interval
  UL=np.percentile(sorted_median_Array,UL_per) # Upper Limit of the confidence interval

  # Return the confidence interval and a descriptive string
  return [LL,UL],f"with {CI}% CI"
```
